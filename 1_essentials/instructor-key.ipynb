### Particle in a 1-Dimensional Box


:::{figure} https://upload.wikimedia.org/wikipedia/commons/thumb/1/13/Infinite_potential_well-en.svg/3840px-Infinite_potential_well-en.svg.png?raw=true
:label: fig:1dpib-potential
:align: center
:width: 500px
:alt: Graph of the infinite square well potential. The horizontal real axis over variable x represents particle position, the vertical axis is particle energy E. From x = negative infinity to x = 0, the potential energy function V of x is equal to positive infinity. This region of the graph is labeled as a "barrier" with the value of V defined as equal to infinity, and is colored solid grey. Between x = 0 and x = L, the graph is labeled as a "well" with V = 0, and the graph is colored white. For values of x larger than L, the graph is again labeled as a "barrier" with the value of V defined to be equal to infinity.

Potential energy function $V(x)$ defining the "box" in which the particle is confined, i.e., $0<x<L$. 
:::


You have already seen that an isolated particle of mass $m$ which is confined to a 1-dimensional "box" of length $L$ defined by the potential energy function
\begin{equation}
V(x) = \begin{cases} 0 & 0\leq x \leq L\\ \infty & {\rm otherwise}\end{cases}
\label{eqn:1dpib-potential}
\end{equation}
(visualized above in [](#fig:1dpib-potential)) is subject to the time-independent wave equation (TIWE)
\begin{equation}
\frac{-\hbar^2}{2m}\frac{d^2}{dx^2}\psi(x) + V(x)\psi(x) = E\psi(x),
\label{eqn:1dpib-tiwe}
\end{equation}


which, when solved by way of integration and after the application of boundary conditions yields the solutions
\begin{equation}
\psi_n(x) = \sqrt{\frac{2}{L}}\sin{\frac{n\pi x}{L}};\;\;\; n = 1,\ 2,\ 3,\ \ldots
\label{eqn:1dpib-wfn}
\end{equation}
These _wavefunctions_ $\psi_n(x)$ define the "amplitude" of the particle's matter wave, and $E_n$ is the total energy of the particle when described by a particular wavefunction with definite value of the quantum number $n$. The first postulate of quantum mechanics states that the wavefunction $\psi_n(x)$ contains all knowable information about the particle and its properties, but how can we ask questions of this all-knowing oracle?

### Motivation: Operator Form of the TIWE


It is possible to re-write the TIWE for this system by "factoring out" the wavefunction $\psi_n(x)$ from the left-hand side of [Eqn. %s](#eqn:1dpib-tiwe):
\begin{equation}
\left[\frac{-\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)\right]\psi_n(x) &= E_n\psi_n(x)
\label{eqn:1dpib-tiwe-factoredLHS}
\end{equation}



Now let us define the _Hamiltonian operator_, $\hat{H}$, to represent the mathematical operations contained in the brackets on the left-hand-side of Eqn.[](#eqn:1dpib-tiwe-factoredLHS),
\begin{equation}
\hat{H} = \frac{-\hbar^2}{2m}\frac{d^2}{dx^2} + V(x),
\label{eqn:1dipb-hamiltonian}
\end{equation}
such that our TIWE now becomes
\begin{equation}
\hat{H}\psi_n(x) = E_n\psi_n(x).
\label{eqn:1dpib-Heigeneqn}
\end{equation}


At first, [Eqn. %s](#eqn:1dpib-Heigeneqn) seems to be a trivial change to the TIWE because it seems that $\hat{H}$ contains exactly the same transformations of the wavefunction $\psi_n(x)$ that were present in [Eqn. %s](#eqn:1dpib-tiwe). Rather than the original form of the TIWE which just reads like a differential equation, however, the two sides of the new form of the TIWE in [Eqn. %s](#eqn:1dpib-Heigeneqn) actually read like a conversation:

| Mathematical Notation   | English Translation |
|:------------------------|:--------------------|
| $\hat{H}\psi_n(x)$      | The Hamiltonian operator asks "Hey wavefunction, what's the energy of the particle?" |
| $E\psi_n(x)$            | The wavefunction responds, "It's $E_n$!" |


The power of this form of the TIWE therefore comes from recognizing that the Hamiltonian operator, $\hat{H}$, is how we can ask the question "What is the energy of the system?" of the all-knowing oracle that is the wavefunction $\psi_n(x)$. 


As it turns out there are many more such question-asking operators in quantum mechanics, and the purpose of this lesson is to explore the behavior of these operators within the context of the 1-dimensional "particle in a box" model system.

## I. What is a quantum mechanical operator?


In general, operators (denoted by a capital letter wearing a "hat", e.g., $\hat{\Omega}$) represent a set of instructions to transform a function, vector, or scalar (constant) number into something else by **"acting" to the right**:
$$\hat{\Omega}f = g$$
In this expression, we have made no assumptions about what any of $\hat{\Omega}$, $f$, or $g$ actually look like! In the introduction above, we saw that quantum mechanical operators represent questions that are asked of the wavefunction describing the system at hand. So, what kinds of operators are there in quantum mechanics?


:::{table} Quantum mechanical operators corresponding to the classically observable properties of a model system comprised of a single, isolated particle confined to move in 1 dimension, as well as the question each operator asks of the wavefunction $\psi(x)$. Table adapted from Tbl. 4.1 of McQuarrie & Simon.
:label: tbl:1d-qm-operators
:align: center
| QM Operator | Question Asked | Answer: Measured Classical Observable |
|:------------|:---------------|:-------------------------------------:|
| $\hat{X} = x\times\;\;\;\;$ (multiply to the right by $x$) | Where is the particle? | Scalar position, $x$ |
| $\hat{P}_x = -i\hbar\frac{d}{dx}$ | What is the particle's momentum? | Linear momentum in $x$ direction, $p_x$ |
| $\hat{T}_x = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2}$ | What is the particle's kinetic energy? | Linear kinetic energy in $x$ direction, $T_x$ |
| $\hat{V}_x = V(x)\times\;\;\;\;$ | What is the particle's potential energy? | Potential energy in $x$ direction, $V(x)$ |
| $\hat{H} = \hat{T}_x + \hat{V}_x = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)\times$ | What is the particle's total energy? | Total energy, $E$ |
:::


In quantum mechanics there are many more operators than are listed above, however for the purpses of this lesson we will only concern ourselves with those defined in [](#tbl:1d-qm-operators) because they are the most relevant for our 1-dimensional model system.

### Property 1: QM Operators Act to the Right


Unlike the mathematical notation $\hat{\Omega} f(x) = g(x)$ may suggest, the action of an operator to the right onto a function is not necessarily limited to multiplication. Indeed, three of the QM operators defined in [](#tbl:1d-qm-operators) involve something _other_ than multiplication! So, how do we apply an operator to the right in SymPy? 


::::{tip}🐍 Manually Applying Operators in SymPy
:icon:false

$$\hat{D}_x f(x) = \frac{d}{dx}\left[x^2 + 2x - 3\right] = 2x + 2,$$


### Property 2: Operator Products Act "Inside Out"


As you saw in the above example, it is possible to apply operators sequentially to a function. In fact, this is an illustration of an _operator product_, where each operator is applied to the previous result in an "inside out" fashion:
$$\hat{P}_x\hat{X}\hat{P}_x \psi_n(x) = \hat{P}_x\left\{\hat{X}\left[\hat{P}_x\psi_n(x)\right]\right\}$$


While it is possible to collect all of the transformations made by an operator product into a single "compound" operator which can be applied all at once, doing so correctly requires more effort than simply applying the operators in the product to each other and is beyond the scope of this lesson. Instead, it is the author's recommendation to apply an operator product in the same manner that was scaffolded in the previous example: one step at a time, by assigning intermediate results to a Python variable that is then acted upon by the next operator in line.

:::{tip} 💸 Strategic Tip: Apply Operator Products Sequentially
:icon:false
:class:dropdown
For the sake of code readability and avoiding unnecessary errors, apply operator products in an inside-out, stepwise fashion by assigning each intermediate result to a Python variable that is acted upon by the next operator in sequence.

For example, to evaluate the operator product
$$\hat{D}\hat{C}\hat{B}\hat{A} f(x),$$ 
1. Apply the innermost operator $\hat{A}$ to the function $f(x)$ and assign the result to the intermediate variable `A_f`
2. Apply the next outer operator $\hat{B}$ to the previous result `A_f`, assign the new result to the intermediate variable `B_A_f`
3. Apply the next outer operator $\hat{C}$ to the previous result `B_A_f`, assign the new result to the intermediate variable `C_B_A_f`
4. Apply the outermost operator $\hat{D}$ to the previous result `C_B_A_f`, assign the new result to the intermediate variable `D_C_B_A_f`
:::

### Property 3: Powers of Operators Are Just Products


Because operator products are evaluated sequentially, some expressions involving operator products can become kinda...wide. For example, the expression for the fourth derivative of a function could be written as
\begin{equation*}
f^{(4)}(x) = \hat{D}_x\hat{D}_x\hat{D}_x\hat{D}_x f(x) = \frac{d}{dx}\left[\frac{d}{dx}\left[\frac{d}{dx}\left[\frac{d}{dx} f(x)\right]\right]\right]
\end{equation*}
To make operator expressions a bit more readable we borrow the "power notation" from algebra (e.g., $x^3 = x\cdot x\cdot x$), and use the exponent of an operator to denote how many times in a row it is applied to its target.

:::{note}ℹ️ Definition: Raising an Operator to a Power
:icon:false
An operator raised to a positive, whole-number power $n$ corresponds to the successive application of that operator $n$ times, i.e.,
$$\hat{O}^n f = \hat{O}\left[\hat{O}\left[\hat{O}\left[\ldots\left[\hat{O} f\right]\right]\right]\right].$$
:::


Using this notation is already equivalent to the way we write multiple derivatives in calculus, making the operator version of the fourth derivative expression above read as
\begin{equation*}
f^{(4)}(x) = \hat{D}^4_x f(x) = \frac{d^4}{dx^4} f(x)
\end{equation*}

### Property 4: QM Operators are Linear


We have seen that operators act to transform functions, and products of operators can be evaluated by applying each operator sequentially in an inside-out manner. In both of the above examples illustrating these properties, we have applied operators to expressions with a single term (as opposed to multiple terms added together). This seems reasonable, as we have been operating on the wavefunction given in [Eqn. %s](#eqn:1dpib-wfn), which only has one term. 


It is a common occurrence, however, where the wavefunction for the system at hand is _not_ describable as exactly one of the solutions to the TIWE for that system. For example, consider a particle in a 1-dimensional box whose wavefunction is actually a linear combination of the first two TIWE solutions, i.e., $\psi_1(x)$ (for $n = 1$) and $\psi_2(x)$ (for $n = 2$):
\begin{equation}
\phi(x) = c_1\psi_1(x) + c_2\psi_2(x);\;\;\;\; \begin{cases} 0 < c_1 < 1\\ 0 < c_2 < 1\end{cases}
\end{equation}
where the positive, nonzero real numbers $c_1,\ c_2$ are called _expansion coefficients_. How could we apply the linear momentum operator $\hat{P}_x$ to this wavefunction?

:::{note}ℹ️ QM Operators are Linear
:icon:false
All quantum mechanical operators are _linear_, meaning they distribute to the right into a sum in the normal way:
$$\hat{\Omega} \left(a f(x) + b g(x)\right) = a\hat{\Omega}f(x) + b\hat{\Omega}g(x)$$
Furthermore, functions being acted upon by a sum of linear operators distribute to the left into the operator expression:
$$\left(a\hat{\Omega} + b\hat{\Lambda}\right) f(x) = a\hat{\Omega}f(x) + b\hat{\Lambda}f(x).$$
Expressions involving a sum of linear operators acting to the right onto a sum of functions are therefore said to be _distributive_:
$$\left[a\hat{\Omega} + b\hat{\Lambda}\right]\left(i f(x) + j g(x)\right) = ai\hat{\Omega} f(x) + aj\hat{\Omega} g(x) + bi\hat{\Lambda} f(x) + bj\hat{\Lambda} g(x),$$
or more generally,
$$\left[\sum_{n}a_n\hat{A}_n\right] \left(\sum_{i} c_i f_i(x)\right) = \sum_n\sum_i a_nc_i\hat{A}_nf_i(x).$$
:::

### Property 5: Commutativity Not Guaranteed


Standard arithmetic multiplication is _commutative_, i.e., for numbers $a$ and $b$ it is true that
$$ab = ba$$
Because operators act directionally, however, it is not generally true that operators will _commute_ when applied to an arbitrary function $f(x)$, i.e.,
\begin{equation}
\hat{\Omega}\hat{\Lambda} f(x) \stackrel{?}{=} \hat{\Lambda}\hat{\Omega} f(x).
\label{eqn:commutes?}
\end{equation}
If both sides of [Eqn. %s](#eqn:commutes?) are equal, then the operators $\hat{\Omega}$ and $\hat{\Lambda}$ are said to _commute_ or to _be commutative_, however if they are not equal the operators are said to be _noncommutative_.

## II. QM Operators & Measurement


At the beginning of this lesson, we motivated our interest in quantum mechanical operators by the fact that the time-independent wave equation can be rewritten from its differential form in [Eqn. %s](#eqn:1dpib-tiwe) to the operator equation [Eqn. %s](#eqn:1dpib-Heigeneqn), where the Hamiltonian operator $\hat{H}$ asks a question about the energy of the particle described by the wavefunction $\psi_n(x)$. Furthermore, we know that to every classically observable property of a system, there corresponds a quantum mechanical operator whose question relates to that observable property.

In this way, a quantum mechanical operator is fundamentally tied to the _measurement_ of its classically observable property. As with any experiment it is always good practice to predict the results of a measurement as well as to interpret the significance of its results, so how can we use quantum mechanical operators to do this?

### Expected Value of a Measurement


For a quantum mechanical system described by the general wavefunction $\Phi(\tau)$, the average value of classically observable property $\omega$ corresponding to the QM operator $\hat{\Omega}$, denoted $\langle\omega\rangle$, is given by
\begin{equation}
\langle\omega\rangle = \frac{\underset{\rm all\ space}{\displaystyle\int}d\tau\ \Phi^{\ast}(\tau)\hat{\Omega}\Phi(\tau)}{\underset{\rm all\ space}{\displaystyle\int}d\tau\ \Phi^{\ast}(\tau)\Phi(\tau)}
\label{eqn:operator-expectation-unnormed}
\end{equation}

where 
* the integrals in the expression above are evaluated over "all space," i.e., for all possible values taken on by the collective independent variable $\tau$, and
* The asterisk on the terms $\Phi^{\ast}(\tau)$ in both integrands denotes taking the _complex conjugate_ of the wavefunction by inverting the sign of the imaginary number $i$ wherever present in $\Phi(\tau)$.

The quantity $\langle\omega\rangle$ is also known as the _expectation value of $\hat{\Omega}$_, denoted $\langle\hat{\Omega}\rangle$, because it is the value which can be expected from an experimental measurement of the classical observable $\omega$.[^fn:expected-vs-actual-measurement]

[^fn:expected-vs-actual-measurement]: It should be noted here that this _expected value_ $\langle\omega\rangle$ of a measurement is **almost never** the same as the _actual value_ which is observed from a single measurement itself. Rather, it is what would result if the observed value from infinitely many such measurements of infinitely many identical systems were averaged together.

As we will see below in Section III, one property of every wavefunction $\Phi(\tau)$ which solves the TIWE is that no matter if they are initially normalized, there always exists a _normalization constant_ $N$ for which

\end{equation*}
such that the wavefunction can be _normalized_ as $\Psi(\tau) = N\Phi(\tau)$. To avoid needing to ever use [Eqn. %s](#eqn:operator-expectation-unnormed), let's briefly discuss how to normalize a wavefunction in SymPy.

### Measurement Uncertainty


After a sufficient number of experimental trials measuring a classically observable property $\omega$ have been performed, the observed values will eventually average out to the expectation value $\langle\omega\rangle = \langle\hat{\Omega}\rangle$. As with any experiment, random error and noise will always result in values being measured which deviate from this average, an effect which is also impacted by the inherently probabilistic nature of quantum mechanical systems. 

How then can we predict this variance in our observation?

:::{note}ℹ️ Definition: Variance in an Operator Expectation Value
:icon:false
For a quantum mechanical system described by a wavefunction $\Psi(\tau)$, the _variance_ in the measurement of a classically observable property $\omega$ corresponding to the quantum mechanical operator $\hat{\Omega}$ is given by
\begin{equation}
\sigma^2_{\omega} = \left(\Delta\hat{\Omega}\right)^2 = \langle\hat{\Omega}^2\rangle - \langle\hat{\Omega}\rangle^2
\label{eqn:operator-variance}
\end{equation}
Similarly, the _standard uncertainty_ (or _standard deviation_) in the expected value is given by
\begin{equation}
\sigma_{\omega} = \Delta\hat{\Omega} = \sqrt{\sigma^2_{\omega}} = \left[\langle\hat{\Omega}^2\rangle - \langle\hat{\Omega}\rangle^2\right]^{1/2}
\label{eqn:operator-stdev}
\end{equation}
:::

## III. The TIWE as an Eigenvalue Equation



At the beginning of the lesson, we motivated our interest in quantum mechanical operators by rewriting the time-independent wave equation from its differential form in [Eqn. %s](#eqn:1dpib-tiwe) to the operator equation of [Eqn. %s](#eqn:1dpib-Heigeneqn), reproduced below as
\begin{equation}
\hat{H}\psi_n(x) = E_n\psi_n(x)
\label{eqn:1dpib-Heigeneqn}
\end{equation}
where the quantum number $n=1,2,3,\ldots$ has been added to index the set of energies $\{E_n\}$ and wavefunctions $\{\psi_n(x)\}$. Unlike the examples above where $\hat{X}$ and $\hat{P}_x$ both acted upon $\psi_n(x)$ to transform it into something else entirely, each of the wavefunctions $\{\psi_n(x)\}$ share a special relationship with the Hamiltonian operator $\hat{H}$: the action of $\hat{H}$ onto any $\psi_n(x)$ returns a particular constant, $E_n$, but otherwise leaves the $\psi_n(x)$ unchanged. 


This form of the TIWE is a special type of equation called an _eigenvalue equation_, where 
* the set of wavefunctions $\{\psi_n(x)\}$ are called the _eigenfunctions_ or _eigenstates_ of the Hamiltonian operator $\hat{H}$, and
* the set of energies $\{E_n\}$ are called the _eigenvalues_ of $\hat{H}$, with each eigenvalue $E_n$ corresponding to the particular eigenstate $\psi_n(x)$.

:::{note} ℹ️ "Eigen"functions & "Eigen"values
:icon:false
The German word "eigen" translates as "own," i.e., the eigenstates of an operator are the operator's own functions, and the eigenvalue associated with the eigenstate is that function's own value.
:::

### II.B. Operators with Simultaneous Eigenstates



In section I.E above, we saw that operators are often noncommutative, such that their commutator $\left[\hat{\Omega},\hat{\Lambda}\right]\not\equiv\hat{\varnothing}$. For operators who _do_ commute, however (i.e., those for whom $\left[\hat{\Omega},\hat{\Lambda}\right]\equiv\hat{\varnothing}$), an interesting relationship between their eigenstates follows:

:::{note} ℹ️ Operator Eigenstates & Commutivity
:icon:false
If two linear operators $\hat{\Omega}$, $\hat{\Lambda}$ commute, i.e., $\left[\hat{\Omega},\hat{\Lambda}\right] \equiv \hat{\varnothing}$, they will share a set of simultaneous eigenstates $\left\{\phi_n\right\}$ such that
\begin{align}
\hat{\Omega}\phi_n &= \omega_n\phi_n\\
\hat{\Lambda}\phi_n &= \lambda_n\phi_n,
\end{align}
and where the two sets of eigenstates $\left\{\lambda_n\right\}$ and $\left\{\omega_n\right\}$ differ only up to a constant factor:
$$\lambda_n = {\rm constant}\times\omega_n$$
:::

### III.B. Properties of QM Operators & Eigenstates

$$\hat{H} = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2}$$

#### III.B.1 Eigenstate Normalizability

:::{note}ℹ️ Property III.B.1: Eigenstates of Hermitian Operators are Normalizable
:icon:false
The eigenstates $\left\{\phi_n\right\}$ of a Hermitian operator $\hat{\Omega}$ are _normalizable_, meaning that it is always possible to find a multiplicative constant $N^2$ such that the overlap integral
\begin{equation}
\label{eqn:normalization-integral}
N^2\int_{\mathbb{R}^3} d^{3}{\bf r}\ \phi_n^{\ast}({\bf r})\phi_n({\bf r}) = 1,
\end{equation}
where the asterisk (${}^{\ast}$) represents the complex conjugate operation. In other words, it is always possible to find a _normalization constant_ $N$ such that for $\psi_n(x) = N\cdot\phi_n(x)$,
$$1 = \int_{\mathbb{R}^3} d^{3}{\bf r}\ \psi_n^{\ast}({\bf r})\psi_n({\bf r}).$$
:::

#### III.B.2 Eigenstate Orthonormality

:::{note} ℹ️ Property III.B.2: Eigenstates of Hermitian Operators are Orthonormal
:icon:false
The eigenstates $\left\{\phi_n({\bf r})\right\}$ of a Hermitian operator $\hat{\Omega}$ can be chosen to be mutually orthogonal, meaning that their overlap integral
$$\int_{\mathbb{R}^3} d^{3}{\bf r}\ \phi^{\ast}_m({\bf r})\phi_n({\bf r}) = 0\ {\rm for\ }m\neq n.$$
Combined with Property III.B.1 above that all eigenstates of a Hermitian operator are normalizable, this means that the eigenstates of a Hermitian operator can be chosen to be _orthonormal_, such that the overlap integral between normalized eigenstates $\psi_n({\bf r}) = N\phi_n({\bf r})$ becomes 
$$\int_{\mathbb{R}^3} d^{3}{\bf r}\ \psi^{\ast}_m({\bf r})\psi_n({\bf r}) = \delta_{mn}= \begin{cases}0 & m\neq n\\ 1 & m=n\end{cases},$$
where $\delta_{mn}$ is called the _Kronecker delta function_.

#### III.B.4 Degenerate Eigenstates



:::{note}ℹ️ Definition: Eigenstate Degeneracy
:icon:false
Two eigenstates $\phi_1$, $\phi_2$ of a linear operator $\hat{\Omega}$ are said to be _degenerate_ if they share an eigenvalue, i.e., 
\begin{align}
\hat{\Omega}\phi_1 &= \omega\phi_1\\
\hat{\Omega}\phi_2 &= \omega\phi_2
\end{align}
Every linear combination of these degenerate eigenstates, e.g. $c_1\phi_1 + c_2\phi_2$, will also be an eigenstate of the operator $\hat{\Omega}$, sharing the same eigenvalue $\omega$ as its constituent eigenstates:
$$\hat{\Omega}\left(c_1\phi_1 + c_2\phi_2\right) = c_1\hat{\Omega}\phi_1 + c_2\hat{\Omega}\phi_2 = c_1\omega\phi_1 + c_2\omega\phi_2 = \omega\left(c_1\phi_1 + c_2\phi_2\right)$$
:::

#### III.C.1 Expectation Value of an Observable



:::{note}ℹ️ Expectation Values of Observables & Operators
:icon:false
Another postulate of quantum mechanics states that if a system is described by a normalized wavefunction $\Psi(\tau)$, then a measurement of the classical observable $\omega$ corresponding to the QM operator $\hat{\Omega}$ will have the _expected value_ given by
$$\langle\ \omega\ \rangle = \langle\ \hat{\Omega}\ \rangle = \underset{\rm all\ space}{\int}d\tau\ \Psi^{\ast}(\tau)\hat{\Omega}\Psi(\tau),$$
where $\tau$ represents the collective variables of all particles within the QM system and the integration is performed over all possible values of this collective variable (i.e, "all space").

:icon:false
If the system is in an eigenstate $\psi_n(\tau)$ of the operator $\hat{\Omega}$, then the variance in a measurement of the classical observable $\omega$ is
\begin{align}
\sigma^2_{\omega} &= \langle\ \hat{\Omega}^2\ \rangle - \langle\ \hat{\Omega}\ \rangle^2\\
&= \underset{\rm all\ space}{\int}d\tau\ \psi_n^{\ast}(\tau)\hat{\Omega}^2\psi_n(\tau) - \left[\underset{\rm all\ space}{\int}d\tau\ \psi_n^{\ast}(\tau)\hat{\Omega}\psi_n(\tau)\right]^2\\

&= \omega_n^2\left[\underset{\rm all\ space}{\int}d\tau\ \psi_n^{\ast}(\tau)\psi_n(\tau)\right] - \omega_n^2\left(1\right)^2\\
&= \omega_n^2\left(1\right)^2 - \omega_n^2\\
\therefore \sigma^2_{\omega} &= 0.
\end{align}

### III.D. Operator Uncertainty Relationships



One of the most famous results from quantum mechanics is the so called _Heisenberg Uncertainty Principle_, which states that the position and momentum for a quantum mechanical particle cannot be known simultaneously to infinite precision, and instead obey the relationship
$$\sigma_x\sigma_{p_x} \geq \frac{\hbar}{2}$$
where $\sigma_x = \sqrt{\sigma^2_x}$ and $\sigma_{p_x} = \sqrt{\sigma^2_{p_x}}$ represent the _standard uncertainty_ (a.k.a. the _standard deviation_) in the particle's position and momentum, respectively. Though famous, this particular uncertainty relationship is not special; in fact, many such uncertainty relationships exist between classical observables of quantum mechanical systems. Most notable about these uncertainty relationships is the fact that _they are inherent to the formulation of quantum mechanics itself_, rather than resulting from inaccuracies in measurement or poorly performed experiments. So what about quantum mechanics gives rise to uncertainty relationships?

:::{note}ℹ️ Definition: Uncertainty Relationship Between Classical Observables for a QM System
:icon:false
For a quantum mechanical system described by state function $\Psi(\tau)$, the two classical observables $\lambda$, $\omega$ corresponding to the linear Hermitian operators $\hat{\Lambda}$, $\hat{\Omega}$ will have the uncertainty relationship
$$\sigma_{\lambda}\sigma_{\omega} \geq \frac{1}{2}\left\vert\underset{\rm all\ space}{\int} d\tau\ \Psi^{\ast}(\tau)\left[\hat{\Lambda},\hat{\Omega}\right]\Psi(\tau)\right\vert,$$
where the vertical bars indicate the complelx modulus (a.k.a. absolute value) of the integral.
:::
