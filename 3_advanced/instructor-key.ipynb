## Introduction


In 1925-1926, two independent formulations of quantum mechanics were introduced. Most undergraduate physical chemistry courses first introduce the so-called _wave mechanics_ formulation, which is concerned with solving the time-independent wave equation (TIWE)
$$\frac{\hbar}{2m}\nabla^2\Psi(x,y,z) + V(x,y,z)\cdot\Psi(x,y,z) = E\cdot\Psi(x,y,z)$$
for the _wavefunction_ $\Psi(x,y,z)$, which describes the ``amplitude'' in three dimensional space of a single particle of mass $m$ subject to the external potential $V(x,y,z)$. The TIWE was immediately succesful for describing single-particle systems that had previously stymied classical physics (e.g., the atomic structure and line spectrum of the Hydrogen atom), however it becomes impossible to solve for more than one particle and this formulation offers no clear path forward for such systems.


The second formulation of quantum mechanics, called _matrix mechanics_, instead writes the TIWE in terms of operators and matrices. While this formulation provides deep physical insight into the properties of quantum mechanical systems under consideration and indeed quantum mechanics itself, working with operators can be a significant conceptual leap for students in an undergraduate course in physical chemistry. The purpose of this lesson is therefore to help students apply the SymPy library to both explore the properties of operators themselves and to apply operators within a quantum mechanical context.

## I. General Operator Properties


In general, operators (denoted by a capital letter wearing a "hat", e.g., $\hat{\Omega}$) represent a set of instructions to transform a function, vector, or scalar (constant) number into something else by **"acting" to the right**:
$$\hat{\Omega}f = g$$
In this expression, we have made no assumptions about what any of $\hat{\Omega}$, $f$, or $g$ actually look like! Some possibilities include the following:


| Operator Definition         | Action on constant $c$, variable $x$, or function $f(x)$ | English translation                              |
|-----------------------------|----------------------------------------------------------|--------------------------------------------------|
| $\hat{X} := x\cdot$         | $\hat{X}f(x) = x\cdot f(x)$                              | Multiply to the right by $x$                     |
| $\hat{D}_x := \frac{d}{dx}$ | $\hat{D}_x f(x) = \frac{d}{dx}f(x) = f'(x)$              | Differentiate $f(x)$ with respect to $x$         |
| ${\rm SQRT} := \sqrt{\;\;}$ | ${\rm SQRT} f(x) = \sqrt{f(x)}$                          | Take the square root of $f(x)$                   |


Because operators represent a set of instructions for transforming one object into another, let's begin by defining some scalars and functions using SymPy that we will use to explore operators and their properties below.

### I.A. Coding Operator Expressions

Unlike the mathematical notation $\hat{\Omega} f(x) = g(x)$ may suggest, operators do not necessarily only act by "multiplying" to the right. For example, consider the $function f(x) = \sin{x}$. While the operation 
$$\hat{Z}f(x) = z\cdot\sin{x}$$ 
_does_ just involve multiplication, $\hat{D}_z f(x)$ does not:
\begin{equation}
\hat{D}_z g(x) = \frac{\partial}{\partial z} \sin{x} = 0
\end{equation}


Particularly when applying operators to mathematical objects using SymPy, it is important that the person coding the expression first translate the definition of the operator to its Python/SymPy representation, and only then perform that operation on the object of the operator's action. For the example above, you must actually evaluate the $z$-derivative of the function $g(x)$ using the SymPy `diff()` function:
```{code}python
:linenos:
# Apply Dz = d/dz operator onto g(x) = sin(x):
Dz_g = diff(g,z)
```
In other words, whenever applying your operator would require calling a Python or SymPy function, _you must actually apply the operator **yourself**._

~~~{hint}🐍 Operator Expressions Calling Python/SymPy Functions
:icon:false
When applying an operator to a SymPy object would require calling a Python or SymPy function, you must be the one to apply the operator by calling the appropriate functions on the object of the operator's action.

For example, for the operator $\hat{\Omega} = -\hat{D}_x^2 + \frac{1}{2}\hat{X}^2= -\frac{\partial^2}{\partial x^2} + \frac{1}{2}x^2$ and function $g(x) = \sin(x)$, the operator expression 
$$\hat{\Omega}g(x) = -\frac{\partial^2}{\partial x^2} \sin{x} + \frac{1}{2}x^2\sin{x}$$

### I.B. Operator Linearity

:::{note}ℹ️ Definition of a Linear Operator
:icon:false
An operator is said to be _linear_ if it distributes into a sum in the normal way:
$$\hat{\Omega} \left(a f(x) + b g(x)\right) = a\hat{\Omega}f(x) + b\hat{\Omega}g(x),$$
or more generally,
$$\hat{\Omega} \left(\sum_{i} c_i f_i(x)\right) = \sum_i c_i\hat{\Omega}f_i(x).$$
:::

In addition to the fact that a linear operator will distribute into a sum in the normal way, functions can also distribute into expressions of linear operators. Consider the following expression involving linear operators $\hat{\Omega}$ and $\hat{\Lambda}$:
$$\left(\hat{\Omega} + \hat{\Lambda}\right)f = \hat{\Omega}f + \hat{\Lambda}f$$

:::{tip}👀 Point of Emphasis: Directionality of Operator Expressions
:icon:false
Because operators act **to the right**, the "normal" mathematical behavior that we expect --- i.e., that left-to-right and right-to-left arithmetic operations are equivalent --- is no longer guaranteed to be true! For example,
\begin{align}
\hat{\Omega}f &\neq f\hat{\Omega}\\
\left(\hat{\Omega} + \hat{\Lambda}\right)f = \hat{\Omega}f + \hat{\Lambda}f &\neq f\hat{\Omega} + f\hat{\Lambda} = f\left(\hat{\Omega} + \hat{\Lambda}\right)
\end{align}
:::

### I.C. Products of Operators

:::{note}ℹ️ Definition: Products of Operators
:icon:false
Operators in a product are applied from the inside-out, in order of proximity to their object of operation:
$$\hat{\Omega}\hat{\Lambda} f(x) = \hat{\Omega}\left[\hat{\Lambda} f(x)\right] = \hat{\Omega} g(x) = h(x)$$
To evaluate expressions involving products of operators, it is essential to evaluate the result of each operation in the order it appears in the expression.
:::


For example, consider the following expression where the operators $\hat{\Omega} = \frac{d^2}{dx^2} + kx^2$ and $\hat{\Lambda} = \frac{d}{dy}$ act on the real-valued scalar function $f(x,y) = ye^{ax^2}$, where $k, a \geq 0\in\mathbb{R}$ are nonnegative real constants:
$$\hat{\Omega}\hat{\Lambda} f(x,y)$$ 

To evaluate this expression, the operator $\hat{\Lambda}$ must first be applied to $f(x,y)$, before $\hat{\Omega}$ is applied to the result:

\begin{align}
\hat{\Omega}\hat{\Lambda}\left(ye^{a x^2}\right) &= \hat{\Omega}\left[\hat{\Lambda}\left(ye^{a x^2}\right)\right]\\
&= \hat{\Omega}\left[\frac{d}{dy}ye^{a x^2}\right]\\
&= \hat{\Omega}e^{a x^2}\\
&= \left(\frac{d^2}{dx^2} + kx^2\right)e^{a x^2}\\
&= \frac{d^2}{dx^2}e^{a x^2} + kx^2 e^{a x^2}\\
&= 2 a \left(2 a x^{2} + 1\right) e^{a x^{2}} + k x^{2} e^{a x^{2}}
\end{align}

#### I.C.2. Closed-Form Expressions of Operator Products


Rather than applying each operator in a product in a stepwise fashion, it is also possible to construct a closed-form expression 
$$\hat{\Gamma} = \hat{\Omega}\hat{\Lambda}$$ 
which transforms its target in a single step. To construct this compound operator, the action of the operator product on an _arbitrary_ function $F$ must first be determined. Then, the arbitrary function can be collected and removed at the end to leave only the closed-form expression for $\hat{\Gamma}$.


To illustrate this process, consider the operator product $\hat{\Gamma} = \hat{D}_x\hat{X}$. It is tempting to construct $\hat{\Gamma}$ by simply applying $\hat{D}_x$ onto the $\hat{X}$ operator:
$$\hat{\Gamma} = \hat{D}_x\hat{X} = \frac{\partial}{\partial x} x\cdot = 1\cdot,$$
which would imply that $\hat{\Gamma} = 1\cdot$ is the _identity operator_ which leaves its target unchanged. The error in this process becomes clear when we instead apply the product $\hat{D}_x\hat{X}$ to the arbitrary function $F = f(x,y,\ldots)$:

\begin{align}
\hat{\Gamma} F = \hat{D}_x\hat{X} F &= \frac{\partial}{\partial x} x\cdot F &&({\rm definintions\ of\ \hat{D}_x\ and\ \hat{X}})\\
&= \frac{\partial}{\partial x} x F &&({\rm apply\ \hat{X}\ onto\ }F)\\
&= x \frac{\partial}{\partial x}F + F\frac{\partial}{\partial x} x &&({\rm product\ rule})\\
&= x \frac{\partial}{\partial x}F + F &&\left(\frac{\partial}{\partial x} x = 1\right)\\
\hat{\Gamma} F &= \left[x \frac{\partial}{\partial x} + 1\right]F &&({\rm factor\ out\ } F {\rm\ from\ the\ right})\\
\therefore\ \hat{\Gamma} &= {\color{blue}x \frac{\partial}{\partial x}} + 1
\end{align}


By not evaluating the action of the operator product on an arbitrary function $F$, the first approach failed to properly account for the product rule of differentiation and completely missed the additional term $x\frac{\partial}{\partial x}$ highlighted in red in the expression above! 

:::{danger}☢️ Critically Important
:icon:false
Evaluating the action of the operator product on an arbitrary function is _critically_ important because it is otherwise possible to mistakenly cancel terms that are actually present in the closed-form expression.

***Always*** include action on an arbitrary function when forming compound operators!
:::

### I.D. Powers of Operators

:::{note}ℹ️ Definition: Raising an Operator to a Power
:icon:false
An operator raised to a positive, whole-number power $n$ corresponds to the successive application of that operator $n$ times, i.e.,
$$\hat{O}^n f = \hat{O}\left[\hat{O}\left[\hat{O}\left[\ldots\left[\hat{O} f\right]\right]\right]\right].$$
:::


Using the notation that raising an exponent to a power $n$ represents applying that operator $n$ times in succession is actually equivalent to the notation we use to write multiple derivatives in calculus. For example, writing the second derivative of the function $x^2$ like
$$\frac{d^2}{dx^2} x^2 = \frac{d}{dx}\left(\frac{d}{dx} x^2\right) = \frac{d}{dx} 2x = 2$$
is exactly the same as using the operator notation
$$\hat{D}_x^2 x^2 = \hat{D}_x\left[\hat{D}_x x^2\right] = \hat{D}_x 2x = 2.$$

This _power notation_ provides a convenient shorthand that can significantly reduce the visual complexity of an operator expression, because $\hat{O}^n f$ is much easier to read and write than the $n$-time-successive-application of the operator via the notation
$$\hat{O}\left[\hat{O}\left[\hat{O}\left[\ldots\left[\hat{O} f\right]\right]\right]\right].$$

Unfortunately, power notation can also be dangerous if we confuse this notation to mean that the _result_ of a single operation is raised to a power rather than the operation being applied $n$ times in succession. To illustrate this, consider the operator expressions $\hat{D}^2_x x^2$ and $\left(\hat{D}_x x^2\right)^2$:
\begin{align}
\hat{D}^2_x x^2 &= \frac{\partial}{\partial x}\left(\frac{\partial}{\partial x} x^2\right) = \frac{\partial}{\partial x} 2x = 2\\
\left(\hat{D}_x x^2\right)^2 &= \left(\frac{\partial}{\partial x}x^2\right)^2 = \left(2x\right)^2 = 4x^2
\end{align}
Clearly $\hat{D}^2_x x^2 \neq \left(\hat{D}_x x^2\right)^2$, and in general this type of equality will only be true in very specific and narrow circumstances.

:::{danger} ☢️ Danger: Operator Power vs. Power of Result
:icon:false
In general, $\hat{O}^n f \neq \left(\hat{O} f\right)^n$.
:::

### I.E. Operator Commutivity


Standard arithmetic multiplication is _commutative_, i.e., for some constants $a$ and $b$ it is true that
$$ab = ba$$
Because operators act directionally, however, it is not generally true that operators will _commute_ when applied to an arbitrary function, i.e.,
$$\hat{\Omega}\hat{\Lambda} F \neq \hat{\Lambda}\hat{\Omega} F.$$ 


Because the action of an operator product on an arbitrary function $F$ in general depends on the order in which the operators are applied, the closed forms for each operator product also depends on the order in which the operator is written, such that
$$\hat{\Omega}\hat{\Lambda} \neq \hat{\Lambda}\hat{\Omega}.$$ 
In order to determine if two operators commute, it is sufficient to test whether the following equality holds:
$$\hat{\Omega}\hat{\Lambda}F - \hat{\Lambda}\hat{\Omega}F \stackrel{?}{=} 0$$


The arbitrary function $F$ can be factored out to the right to define the action of _commutator of $\hat{\Omega}$ and $\hat{\Lambda}$_,
$$\left[\hat{\Omega},\hat{\Lambda}\right]F = \left(\hat{\Omega}\hat{\Lambda} - \hat{\Lambda}\hat{\Omega}\right) F = \hat{\Omega}\hat{\Lambda}F - \hat{\Lambda}\hat{\Omega}F$$


:::{note}ℹ️ Definition: Operator Commutator
:icon:false
The _commutator_ of two operators $\hat{\Omega}$ and $\hat{\Lambda}$, denoted $\left[\hat{\Omega},\hat{\Lambda}\right]$, is given by
\begin{equation}
\left[\hat{\Omega},\hat{\Lambda}\right] = \hat{\Omega}\hat{\Lambda} - \hat{\Lambda}\hat{\Omega}.
\end{equation}

The commutator $\left[\hat{\Omega},\hat{\Lambda}\right]$ is itself an operator, whose action on an arbitrary function $F$ is given by
\begin{equation}
\left[\hat{\Omega},\hat{\Lambda}\right]F = \hat{\Omega}\hat{\Lambda}F - \hat{\Lambda}\hat{\Omega}F.
\end{equation}$$\left[\hat{\Omega},\hat{\Lambda}\right] = \hat{\Omega}\hat{\Lambda} - \hat{\Lambda}\hat{\Omega},$$$$\left[\hat{\Omega},\hat{\Lambda}\right] = \hat{\Omega}\hat{\Lambda} - \hat{\Lambda}\hat{\Omega},$$$$\left[\hat{\Omega},\hat{\Lambda}\right] = \hat{\Omega}\hat{\Lambda} - \hat{\Lambda}\hat{\Omega},$$$$\left[\hat{\Omega},\hat{\Lambda}\right] = \hat{\Omega}\hat{\Lambda} - \hat{\Lambda}\hat{\Omega},$$

## II. Operator Eigenvalue Problems



In general, the operator $\hat{\Omega}$) will act upon a function, vector, or scalar and transform it into something different, i.e.,
$$\hat{\Omega}f = g.$$
Certain operators, however, will have certain objects "of its own" for which
$$\hat{\Omega}\phi(x) = \omega\phi(x),$$
where the object $\phi(x)$ is called an "eigenfunction," "eigenstate," or "eigenvector" of the operator $\hat{\Omega}$ and the constant $\omega$ is called the "eigenvalue" associated with the eigenfunction $\phi(x)$. 

:::{note} ℹ️ "Eigen"functions & "Eigen"values
:icon:false
The German word "eigen" translates as "own," i.e., the eigenstates of an operator are the operator's own functions, and the eigenvalue associated with the eigenstate is that function's own value.
:::

This process is sufficiently involved that the details of solving eigenvalue problems in quantum mechanics will be left for a different lesson. Here, we will instead focus on determining if a given function _is_ an eigenfunction of a particular operator, as well as some general properties of operators and their eigenvalues.

### II.B. Operators with Simultaneous Eigenstates



In section I.E above, we saw that operators are often noncommutative, such that their commutator $\left[\hat{\Omega},\hat{\Lambda}\right]\not\equiv\hat{\varnothing}$. For operators who _do_ commute, however (i.e., those for whom $\left[\hat{\Omega},\hat{\Lambda}\right]\equiv\hat{\varnothing}$), an interesting relationship between their eigenstates follows:

:::{note} ℹ️ Operator Eigenstates & Commutivity
:icon:false
If two linear operators $\hat{\Omega}$, $\hat{\Lambda}$ commute, i.e., $\left[\hat{\Omega},\hat{\Lambda}\right] \equiv \hat{\varnothing}$, they will share a set of simultaneous eigenstates $\left\{\phi_n\right\}$ such that
\begin{align}
\hat{\Omega}\phi_n &= \omega_n\phi_n\\
\hat{\Lambda}\phi_n &= \lambda_n\phi_n,
\end{align}
and where the two sets of eigenstates $\left\{\lambda_n\right\}$ and $\left\{\omega_n\right\}$ differ only up to a constant factor:
$$\lambda_n = {\rm constant}\times\omega_n$$
:::

| Total potential energy, $V(x,y,z)$        | $\hat{V} = V(x,y,z)\cdot,\;\;\;\;{\rm multiply\ to\ the\ right\ by\ } V(x,y,z)$ |
| Total energy, $E$                         | $\hat{H} = \hat{T} + \hat{V} = -\frac{\hbar^2}{2m}\nabla^2 + V(x,y,z)\cdot$ | 
:::

### III.B. Properties of QM Operators & Eigenstates



Beyond the general properties of linear operators we explored above, the additional criterion that all quantum mechanical operators be _Hermitian_ affords additional properties to these operators and their eigenstates that are worth exploring. While these properties are true for all of the Hermitian operators listed in [](#tbl-qm-operators), let us specifically consider a particle subject to a one-dimensional infinite square well potential as an illustrative model system. 


Consider a particle of mass $m$ subject to the potential energy 
$$V(x) = \begin{cases} 0 & 0\leq x\leq\ell\\ \infty & {\rm otherwise}\end{cases}$$
such that the particle is constrained to the region $0\leq x\leq\ell$. Inside this infinite square potential well, he Hamiltonian operator representing the total energy of the particle is given by
$$\hat{H} = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2}$$
where $\hbar = \frac{h}{2\pi}$ is the angular Planck constant. This Hamiltonian defines the eigenvalue equation


$$\hat{H}\psi_n(x) = E_n\psi_n(x),$$
where the eigenstates $\psi_n(x)$ can be shown to be
$$\psi_n(x) = B\sin{\frac{n\pi x}{\ell}};\;\; n\in\mathbb{Z}^+,\ 0\leq B\leq 1\in\mathbb{R}$$
with associated energy eigenvalues
$$E_n = \frac{\hbar^2 n^2 \pi^2}{2m\ell^2}.$$

#### III.B.1 Eigenstate Normalizability

:::{note}ℹ️ Property III.B.1: Eigenstates of Hermitian Operators are Normalizable
:icon:false
The eigenstates $\left\{\phi_n\right\}$ of a Hermitian operator $\hat{\Omega}$ are _normalizable_, meaning that it is always possible to find a multiplicative constant $N^2$ such that the overlap integral
\begin{equation}
\label{eqn:normalization-integral}
N^2\int_{\mathbb{R}^3} d^{3}{\bf r}\ \phi_n^{\ast}({\bf r})\phi_n({\bf r}) = 1,
\end{equation}
where the asterisk (${}^{\ast}$) represents the complex conjugate operation. In other words, it is always possible to find a _normalization constant_ $N$ such that for $\psi_n(x) = N\cdot\phi_n(x)$,
$$1 = \int_{\mathbb{R}^3} d^{3}{\bf r}\ \psi_n^{\ast}({\bf r})\psi_n({\bf r}).$$
:::

#### III.B.2 Eigenstate Orthonormality

:::{note} ℹ️ Property III.B.2: Eigenstates of Hermitian Operators are Orthonormal
:icon:false
The eigenstates $\left\{\phi_n({\bf r})\right\}$ of a Hermitian operator $\hat{\Omega}$ can be chosen to be mutually orthogonal, meaning that their overlap integral
$$\int_{\mathbb{R}^3} d^{3}{\bf r}\ \phi^{\ast}_m({\bf r})\phi_n({\bf r}) = 0\ {\rm for\ }m\neq n.$$
Combined with Property III.B.1 above that all eigenstates of a Hermitian operator are normalizable, this means that the eigenstates of a Hermitian operator can be chosen to be _orthonormal_, such that the overlap integral between normalized eigenstates $\psi_n({\bf r}) = N\phi_n({\bf r})$ becomes 
$$\int_{\mathbb{R}^3} d^{3}{\bf r}\ \psi^{\ast}_m({\bf r})\psi_n({\bf r}) = \delta_{mn}= \begin{cases}0 & m\neq n\\ 1 & m=n\end{cases},$$
where $\delta_{mn}$ is called the _Kronecker delta function_.

#### III.B.4 Degenerate Eigenstates



:::{note}ℹ️ Definition: Eigenstate Degeneracy
:icon:false
Two eigenstates $\phi_1$, $\phi_2$ of a linear operator $\hat{\Omega}$ are said to be _degenerate_ if they share an eigenvalue, i.e., 
\begin{align}
\hat{\Omega}\phi_1 &= \omega\phi_1\\
\hat{\Omega}\phi_2 &= \omega\phi_2
\end{align}
Every linear combination of these degenerate eigenstates, e.g. $c_1\phi_1 + c_2\phi_2$, will also be an eigenstate of the operator $\hat{\Omega}$, sharing the same eigenvalue $\omega$ as its constituent eigenstates:
$$\hat{\Omega}\left(c_1\phi_1 + c_2\phi_2\right) = c_1\hat{\Omega}\phi_1 + c_2\hat{\Omega}\phi_2 = c_1\omega\phi_1 + c_2\omega\phi_2 = \omega\left(c_1\phi_1 + c_2\phi_2\right)$$
:::

### III.D. Operator Uncertainty Relationships



One of the most famous results from quantum mechanics is the so called _Heisenberg Uncertainty Principle_, which states that the position and momentum for a quantum mechanical particle cannot be known simultaneously to infinite precision, and instead obey the relationship
$$\sigma_x\sigma_{p_x} \geq \frac{\hbar}{2}$$
where $\sigma_x = \sqrt{\sigma^2_x}$ and $\sigma_{p_x} = \sqrt{\sigma^2_{p_x}}$ represent the _standard uncertainty_ (a.k.a. the _standard deviation_) in the particle's position and momentum, respectively. Though famous, this particular uncertainty relationship is not special; in fact, many such uncertainty relationships exist between classical observables of quantum mechanical systems. Most notable about these uncertainty relationships is the fact that _they are inherent to the formulation of quantum mechanics itself_, rather than resulting from inaccuracies in measurement or poorly performed experiments. So what about quantum mechanics gives rise to uncertainty relationships?

:::{note}ℹ️ Definition: Uncertainty Relationship Between Classical Observables for a QM System
:icon:false
For a quantum mechanical system described by state function $\Psi(\tau)$, the two classical observables $\lambda$, $\omega$ corresponding to the linear Hermitian operators $\hat{\Lambda}$, $\hat{\Omega}$ will have the uncertainty relationship
$$\sigma_{\lambda}\sigma_{\omega} \geq \frac{1}{2}\left\vert\underset{\rm all\ space}{\int} d\tau\ \Psi^{\ast}(\tau)\left[\hat{\Lambda},\hat{\Omega}\right]\Psi(\tau)\right\vert,$$
where the vertical bars indicate the complelx modulus (a.k.a. absolute value) of the integral.
:::
