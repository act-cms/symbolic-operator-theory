# I. Introduction

## I.A. Particle in a 1-Dimensional Box


:::{figure} https://upload.wikimedia.org/wikipedia/commons/thumb/1/13/Infinite_potential_well-en.svg/3840px-Infinite_potential_well-en.svg.png?raw=true
:label: fig:1dpib-potential
:align: center
:width: 500px
:alt: Graph of the infinite square well potential. The horizontal real axis over variable x represents particle position, the vertical axis is particle energy E. From x = negative infinity to x = 0, the potential energy function V of x is equal to positive infinity. This region of the graph is labeled as a "barrier" with the value of V defined as equal to infinity, and is colored solid grey. Between x = 0 and x = L, the graph is labeled as a "well" with V = 0, and the graph is colored white. For values of x larger than L, the graph is again labeled as a "barrier" with the value of V defined to be equal to infinity.

Potential energy function $V(x)$ defining the "box" in which the particle is confined, i.e., $0<x<L$. 
:::


You have already seen that an isolated particle of mass $m$ which is confined to a 1-dimensional "box" of length $L$ defined by the potential energy function
\begin{equation}
V(x) = \begin{cases} 0 & 0\leq x \leq L\\ \infty & {\rm otherwise}\end{cases}
\label{eqn:1dpib-potential}
\end{equation}
(visualized above in [](#fig:1dpib-potential)) is subject to the time-independent wave equation (TIWE)
\begin{equation}
\frac{-\hbar^2}{2m}\frac{d^2}{dx^2}\psi(x) + V(x)\psi(x) = E\psi(x),
\label{eqn:1dpib-tiwe}
\end{equation}


which, when solved by way of integration and after the application of boundary conditions yields the solutions
\begin{equation}
\psi_n(x) = \sqrt{\frac{2}{L}}\sin{\frac{n\pi x}{L}};\;\;\; n = 1,\ 2,\ 3,\ \ldots
\label{eqn:1dpib-wfn}
\end{equation}
These _wavefunctions_ $\psi_n(x)$ define the "amplitude" of the particle's matter wave, and $E_n$ is the total energy of the particle when described by a particular wavefunction with definite value of the quantum number $n$. The first postulate of quantum mechanics states that the wavefunction $\psi_n(x)$ contains all knowable information about the particle and its properties, but how can we ask questions of this all-knowing oracle?

## I.B. Motivation: Operator Form of the TIWE


It is possible to re-write the TIWE for this system by "factoring out" the wavefunction $\psi_n(x)$ from the left-hand side of [Eqn. %s](#eqn:1dpib-tiwe):
\begin{equation}
\left[\frac{-\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)\right]\psi_n(x) &= E_n\psi_n(x)
\label{eqn:1dpib-tiwe-factoredLHS}
\end{equation}



Now let us define the _Hamiltonian operator_, $\hat{H}$, to represent the mathematical operations contained in the brackets on the left-hand-side of Eqn.[](#eqn:1dpib-tiwe-factoredLHS),
\begin{equation}
\hat{H} = \frac{-\hbar^2}{2m}\frac{d^2}{dx^2} + V(x),
\label{eqn:1dipb-hamiltonian}
\end{equation}
such that our TIWE now becomes
\begin{equation}
\hat{H}\psi_n(x) = E_n\psi_n(x).
\label{eqn:1dpib-Heigeneqn}
\end{equation}


At first, [Eqn. %s](#eqn:1dpib-Heigeneqn) seems to be a trivial change to the TIWE because it seems that $\hat{H}$ contains exactly the same transformations of the wavefunction $\psi_n(x)$ that were present in [Eqn. %s](#eqn:1dpib-tiwe). Rather than the original form of the TIWE which just reads like a differential equation, however, the two sides of the new form of the TIWE in [Eqn. %s](#eqn:1dpib-Heigeneqn) actually read like a conversation:

| Mathematical Notation   | English Translation |
|:------------------------|:--------------------|
| $\hat{H}\psi_n(x)$      | The Hamiltonian operator asks "Hey wavefunction, what's the energy of the particle?" |
| $E\psi_n(x)$            | The wavefunction responds, "It's $E_n$!" |


The power of this form of the TIWE therefore comes from recognizing that the Hamiltonian operator, $\hat{H}$, is how we can ask the question "What is the energy of the system?" of the all-knowing oracle that is the wavefunction $\psi_n(x)$. 


As it turns out there are many more such question-asking operators in quantum mechanics, and the purpose of this lesson is to explore the behavior of these operators within the context of the 1-dimensional "particle in a box" model system.

# II. What is a quantum mechanical operator?


In general, operators (denoted by a capital letter wearing a "hat", e.g., $\hat{\Omega}$) represent a set of instructions to transform a function, vector, or scalar (constant) number into something else by **"acting" to the right**:
$$\hat{\Omega}f = g$$
In this expression, we have made no assumptions about what any of $\hat{\Omega}$, $f$, or $g$ actually look like! In the introduction above, we saw that quantum mechanical operators represent questions that are asked of the wavefunction describing the system at hand. So, what kinds of operators are there in quantum mechanics?

## II.A. QM Operator Definitions


:::{table} Quantum mechanical operators corresponding to the classically observable properties of a model system comprised of a single, isolated particle confined to move in 1 dimension, as well as the question each operator asks of the wavefunction $\psi(x)$. Table adapted from Tbl. 4.1 of McQuarrie & Simon.
:label: tbl:1d-qm-operators
:align: center
| QM Operator | Question Asked | Answer: Measured Classical Observable |
|:------------|:---------------|:-------------------------------------:|
| $\hat{X} = x\times\;\;\;\;$ (multiply to the right by $x$) | Where is the particle? | Scalar position, $x$ |
| $\hat{P}_x = -i\hbar\frac{d}{dx}$ | What is the particle's momentum? | Linear momentum in $x$ direction, $p_x$ |
| $\hat{T}_x = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2}$ | What is the particle's kinetic energy? | Linear kinetic energy in $x$ direction, $T_x$ |
| $\hat{V}_x = V(x)\times\;\;\;\;$ | What is the particle's potential energy? | Potential energy in $x$ direction, $V(x)$ |
| $\hat{H} = \hat{T}_x + \hat{V}_x = -\frac{\hbar^2}{2m}\frac{d^2}{dx^2} + V(x)\times$ | What is the particle's total energy? | Total energy, $E$ |
:::

# III. How do QM Operators Behave?


In quantum mechanics there are many more operators than are listed above, however for the purpses of this lesson we will only concern ourselves with those defined in [](#tbl:1d-qm-operators) because they are the most relevant for our 1-dimensional model system.

## Property 1: QM Operators Act to the Right


Unlike the mathematical notation $\hat{\Omega} f(x) = g(x)$ may suggest, the action of an operator to the right onto a function is not necessarily limited to multiplication. Indeed, three of the QM operators defined in [](#tbl:1d-qm-operators) involve something _other_ than multiplication! So, how do we apply an operator to the right in SymPy? 


::::{tip}🐍 Manually Applying Operators in SymPy

## Property 2: Operator Products Act "Inside Out"


In addition to applying a single operator onto a function, it is also possible to apply operators sequentially to a function. This is an illustration of an _operator product_, where each operator is applied to the previous result sequentially in an "inside out" fashion:
$$\hat{P}_x\hat{X}\hat{P}_x \psi_n(x) = \hat{P}_x\left\{\hat{X}\left[\hat{P}_x\psi_n(x)\right]\right\}$$


While it is possible to collect all of the transformations made by an operator product into a single "compound" operator which can be applied all at once, doing so correctly requires more effort than simply applying the operators in the product to each other and is beyond the scope of this lesson. Instead, it is the author's recommendation to apply an operator product one step at a time, by assigning intermediate results to a Python variable that is then acted upon by the next operator in line.

:::{tip} 💸 Strategic Tip: Apply Operator Products Sequentially
:icon:false
:class:dropdown
For the sake of code readability and avoiding unnecessary errors, apply operator products in an inside-out, stepwise fashion by assigning each intermediate result to a Python variable that is acted upon by the next operator in sequence.

For example, to evaluate the operator product
$$\hat{D}\hat{C}\hat{B}\hat{A} f(x),$$ 
1. Apply the innermost operator $\hat{A}$ to the function $f(x)$ and assign the result to the intermediate variable `A_f`
2. Apply the next outer operator $\hat{B}$ to the previous result `A_f`, assign the new result to the intermediate variable `B_A_f`
3. Apply the next outer operator $\hat{C}$ to the previous result `B_A_f`, assign the new result to the intermediate variable `C_B_A_f`
4. Apply the outermost operator $\hat{D}$ to the previous result `C_B_A_f`, assign the new result to the intermediate variable `D_C_B_A_f`
:::

## Property 3: Powers of Operators Are Just Products


Because operator products are evaluated sequentially, some expressions involving operator products can become kinda...wide. For example, the expression for the fourth derivative of a function could be written as
\begin{equation*}
f^{(4)}(x) = \hat{D}_x\hat{D}_x\hat{D}_x\hat{D}_x f(x) = \frac{d}{dx}\left[\frac{d}{dx}\left[\frac{d}{dx}\left[\frac{d}{dx} f(x)\right]\right]\right]
\end{equation*}
To make operator expressions a bit more readable we borrow the "power notation" from algebra (e.g., $x^3 = x\cdot x\cdot x$), and use the exponent of an operator to denote how many times in a row it is applied to its target.

:::{note}ℹ️ Definition: Raising an Operator to a Power
:icon:false
An operator raised to a positive, whole-number power $n$ corresponds to the successive application of that operator $n$ times, i.e.,
$$\hat{O}^n f = \hat{O}\left[\hat{O}\left[\hat{O}\left[\ldots\left[\hat{O} f\right]\right]\right]\right].$$
:::


Using this notation is already equivalent to the way we write multiple derivatives in calculus, making the operator version of the fourth derivative expression above read as
\begin{equation*}
f^{(4)}(x) = \hat{D}^4_x f(x) = \frac{d^4}{dx^4} f(x)
\end{equation*}

## Property 4: QM Operators are Linear


We have seen that operators act to transform functions, and products of operators can be evaluated by applying each operator sequentially in an inside-out manner. In the examples illustrating these properties, we have applied operators to expressions with a single term (as opposed to multiple terms added together). This seems reasonable, as we have been operating on the wavefunction given in [Eqn. %s](#eqn:1dpib-wfn), which only has one term. 


It is a common occurrence, however, where the wavefunction for the system at hand is _not_ describable as exactly one of the solutions to the TIWE for that system. For example, consider a particle in a 1-dimensional box whose wavefunction is actually a linear combination of the first two TIWE solutions, i.e., $\psi_1(x)$ (for $n = 1$) and $\psi_2(x)$ (for $n = 2$):
\begin{equation}
\phi(x) = c_1\psi_1(x) + c_2\psi_2(x);\;\;\;\; \begin{cases} 0 < c_1 < 1\\ 0 < c_2 < 1\end{cases}
\end{equation}
where the positive, nonzero real numbers $c_1,\ c_2$ are called _expansion coefficients_. How could we apply the linear momentum operator $\hat{P}_x$ to this wavefunction?

:::{note}ℹ️ QM Operators are Linear
:icon:false
All quantum mechanical operators are _linear_, meaning they distribute to the right into a sum in the normal way. For example, consider the action of the arbitrary QM operator $\hat{\Omega}$ on an expression of two functions $f(x)$, $g(x)$ and constants $a,b$:
\begin{equation*}
\hat{\Omega} \left(a f(x) + b g(x)\right) = a\hat{\Omega}f(x) + b\hat{\Omega}g(x)
\end{equation*}
Furthermore, functions being acted upon by a sum of linear operators distribute to the left into the operator expression:
\begin{equation*}
\left(a\hat{\Omega} + b\hat{\Lambda}\right) f(x) = a\hat{\Omega}f(x) + b\hat{\Lambda}f(x).
\end{equation*}
Expressions involving a sum of linear operators acting to the right onto a sum of functions are therefore said to be _distributive_:
\begin{equation*}
\left[a\hat{\Omega} + b\hat{\Lambda}\right]\left(i f(x) + j g(x)\right) = ai\hat{\Omega} f(x) + aj\hat{\Omega} g(x) + bi\hat{\Lambda} f(x) + bj\hat{\Lambda} g(x),
\end{equation*}
or more generally,
\begin{equation}
\left[\sum_{n}a_n\hat{A}_n\right] \left(\sum_{i} c_i f_i(x)\right) = \sum_n\sum_i a_nc_i\hat{A}_nf_i(x).
\label{eqn:operator-linearity}
\end{equation}
:::

:::{tip}👀 Point of Emphasis: Directionality of Operator Expressions
:icon:false
Because operators act **to the right**, the "normal" mathematical behavior that we expect --- i.e., that left-to-right and right-to-left arithmetic operations are equivalent --- is no longer guaranteed to be true! 

For example,
\begin{equation*}
\hat{\Omega}f &\neq f\hat{\Omega}
\end{equation*}
and
\begin{equation*}
\left(\hat{\Omega} + \hat{\Lambda}\right)f = \hat{\Omega}f + \hat{\Lambda}f &\neq f\hat{\Omega} + f\hat{\Lambda} = f\left(\hat{\Omega} + \hat{\Lambda}\right)
\end{equation*}
:::

## Property 5: Commutativity Not Guaranteed


Standard arithmetic multiplication is _commutative_, i.e., for numbers $a$ and $b$ it is true that
$$ab = ba$$
Because operators act directionally, however, it is not generally true that operators will _commute_ when applied to an arbitrary function $f(x)$, i.e.,
\begin{equation}
\hat{\Omega}\hat{\Lambda} f(x) \stackrel{?}{=} \hat{\Lambda}\hat{\Omega} f(x).
\label{eqn:commutes?}
\end{equation}
If both sides of [Eqn. %s](#eqn:commutes?) are equal, then the operators $\hat{\Omega}$ and $\hat{\Lambda}$ are said to _commute_ or to _be commutative_, however if they are not equal the operators are said to be _noncommutative_.
